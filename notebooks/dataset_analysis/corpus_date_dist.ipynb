# Imports

```python
import json
from datetime import datetime
from pathlib import Path
from collections import Counter
```

# Configs

```python
JSONL_FILE = Path(r"C:\Programming\rag_seminar\data-science-seminar\data\raw\financial_markets.jsonl")
TOP_N = 10
```

# Load and Count Publication Dates

```python
date_counts = Counter()

with open(JSONL_FILE, "r", encoding="utf-8") as f:
    for line in f:
        record = json.loads(line)

        published_date = record.get("published_date")

        if not published_date:
            continue

        try:
            # Parse ISO timestamp
            date = datetime.fromisoformat(
                published_date.replace("Z", "+00:00")
            )

            # Count by calendar day
            date_counts[date.date()] += 1

        except ValueError:
            continue

print(f"Found {sum(date_counts.values())} articles")
print(f"Found {len(date_counts)} unique publication dates")
```

# Print Latest Publication Dates

```python
sorted_dates = sorted(
    date_counts.items(),
    key=lambda x: x[0],
    reverse=True
)

print(f"Top {TOP_N} latest publication dates:\n")

for date, count in sorted_dates[:TOP_N]:
    print(f"{date}: {count} articles")
```
