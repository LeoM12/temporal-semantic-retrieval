# Imports

```python
import json
import pandas as pd
```

# Load Results

```python
RESULT_FILE = r"C:\Programming\rag_seminar\data-science-seminar\experiments\testing\queries_evergreen_results.json"

with open(RESULT_FILE, "r", encoding="utf-8") as f:
    data = json.load(f)

print(f"Loaded {len(data)} queries")
```

# Extract Publication Dates

```python
rows = []

for query in data:
    for result in query["results"]:
        rows.append({
            "published_date": result["published_date"]
        })

df = pd.DataFrame(rows)

print(f"Loaded {len(df)} retrieved documents")
df.head()
```

# Calculate age relative to 2025-01-01

```python
REFERENCE_DATE = pd.Timestamp("2025-01-01")

df["published_date"] = pd.to_datetime(df["published_date"])

df["age_days"] = (
    REFERENCE_DATE - df["published_date"]
).dt.days
```

# Mean and median age

```python
mean_age = df["age_days"].mean()
median_age = df["age_days"].median()

print(f"Mean age:   {mean_age:.1f} days")
print(f"Median age: {median_age:.1f} days")
```
